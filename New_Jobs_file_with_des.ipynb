```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the data
df = pd.read_csv('NYC_Jobs_20260713.csv')

# Show the first few rows
df.head()
```

```python
# Show all column names
df.columns

# Show basic info
df.info()
```

```python
# Show top 10 agencies
df['Agency'].value_counts().head(10)
```

```python
# Count internal vs external
df['Posting Type'].value_counts()
```

```python
# Simple pie chart
df['Posting Type'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Internal vs External Postings')
plt.ylabel('')
plt.show()
```

```python
# Show top 10 agencies
df['Agency'].value_counts().head(10)
```

```python
# Bar chart of top agencies
df['Agency'].value_counts().head(10).plot(kind='bar')
plt.title('Top 10 Hiring Agencies')
plt.xlabel('Agency')
plt.ylabel('Number of Postings')
plt.xticks(rotation=45)
plt.show()
```

```python
# Convert to datetime and filter
df['Post Until'] = pd.to_datetime(df['Post Until'])
df_june = df[df['Post Until'].dt.month == 6]

# Show how many rows
len(df_june)
```

```python
# Show top 10 titles in June
df_june['Business Title'].value_counts().head(10)
```

```python
# Plot top titles
df_june['Business Title'].value_counts().head(10).plot(kind='bar')
plt.title('Top 10 Job Titles in June 2026')
plt.xlabel('Job Title')
plt.ylabel('Number of Postings')
plt.xticks(rotation=45)
plt.show()
```

```python
# ============================================================
# CONCLUSION
# ============================================================
print("📢 KEY TAKEAWAYS")
print("=" * 40)
print(f"Total jobs in June 2026: {len(df_june)}")
print(f"External jobs: {len(df_june[df_june['Posting Type'] == 'External'])}")
print(f"Internal jobs: {len(df_june[df_june['Posting Type'] == 'Internal'])}")
print(f"Top agency: {df_june['Agency'].value_counts().index[0]}")
print(f"Top title: {df_june['Business Title'].value_counts().index[0]}")
print("=" * 40)
print("🎯 The city wants to hire. They just don't want to hire us.")
```
