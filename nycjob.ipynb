# DataSet Used : Jobs NYC Posting
#### For our data set we used the dataset that contains the current job postings available on the New York's offical jobs site. 
#### Department of Citywide Administrative Services
#### Last updated June 30, 2026

##### With this data set I am pushing the agenda that it is hard to get a job in NYC government 

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

#starting off we grab our database to read off the categories
df = pd.read_csv("Jobs_NYC_Postings_20260713.csv")
```

```python
#Next we go straight to the posting type. 
#We wanted to find out how are the jobs being seen? 
#Internaly or Externally?

df["Posting Type"].value_counts()
```

```python
#Next we wanted to filter to make sure we only saw jobs around Technology Data and Innovation

filtered_df = df[df["Job Category"].str.contains("Technology, Data & Innovation", na=False)]
sns.countplot(data=filtered_df, x="Posting Type")
```

```python
#Next I wanted to see exactly how many jobs are actually in this category
df["Job Category"].value_counts()
len(df[df["Job Category"].str.contains("Technology, Data & Innovation")])
```

```python
#Then we wanted to cross reference. so if there is 194 jobs in total
#How many are internal and how many are external
tech_df = df[df["Job Category"].str.contains("Technology, Data & Innovation", na=False)]
tech_df["Posting Type"].value_counts()
```

```python
#Still the same. But this isn't pushing my agenda
sns.countplot(data=tech_df, x="Posting Type")
```

```python
#Next I decided to look into Title classification

df["Title Classification"].value_counts()
```

```python
#I wanted to see how in demand these jobs were
#Instead I found out that the title classification of the job is
#how it is determined if a job requires a civil service exam
#Which brings back my agenda because requiring a civil service exam
#is a hurdle when trying to get a job 

order = tech_df["Title Classification"].value_counts().index
sns.countplot(data=tech_df, x="Title Classification", order=order)
plt.xticks(rotation=45, ha='right')
plt.title("Title Classification across all jobs")
plt.ylabel("Number of Postings")
plt.show()
```
