```python
# ============================================================
# Job Hunting Propaganda
# by 2 Gen Z College Students
# ============================================================
# A data-driven look at how NYC is (not) hiring us.
# Dataset: NYC Jobs (June 2025)
# Source: NYC OpenData

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Set style for presentation
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
```

```python
# ============================================================
# 1. LOAD THE DATA
# ============================================================

df = pd.read_csv('NYC_Jobs_20260713.csv')

print(f"📊 Total job postings in June: {len(df)}")
```

```python
# ============================================================
# 2. BASIC OVERVIEW
# ============================================================

# Show first few rows
df.head()

# Check column names
df.columns.tolist()

# Check data types
df.dtypes
```

```python
# ============================================================
# 2. BASIC OVERVIEW
# ============================================================

# Show first few rows
df.head()

# Check column names
df.columns.tolist()

# Check data types
df.dtypes
```

```python
# ============================================================
# 3. POSTING DATE TREND
# ============================================================

# Convert posting date to datetime (if it's not already)
df['Posting Date'] = pd.to_datetime(df['Posting Date'])

# Count postings by day
daily_postings = df['Posting Date'].value_counts().sort_index()

# Plot
plt.figure(figsize=(12, 5))
daily_postings.plot(kind='line', marker='o')
plt.title('📉 NYC Job Postings Over Time (June 2025)')
plt.xlabel('Date')
plt.ylabel('Number of Postings')
plt.grid(True)
plt.show()
```

```python

# ============================================================
# 4. POSTING TYPE DISTRIBUTION
# ============================================================

# Check what values exist
df['Posting Type'].value_counts()

# Plot
df['Posting Type'].value_counts().plot(kind='bar', color=['#3498db', '#e74c3c'])
plt.title('📋 Job Postings by Type')
plt.xlabel('Posting Type')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```

```python

# ============================================================
# 5. NUMBER OF POSITIONS AVAILABLE
# ============================================================

# Summary stats for number of positions
df['# Of Positions'].describe()

# Filter to postings with multiple positions
multi_positions = df[df['# Of Positions'] > 1]
print(f"📌 Postings with multiple positions: {len(multi_positions)}")
```

```python

# ============================================================
# 6. TOP AGENCIES HIRING
# ============================================================

top_agencies = df['Agency'].value_counts().head(10)
top_agencies.plot(kind='bar', color='#2ecc71')
plt.title('🏛️ Top 10 Hiring Agencies (June 2025)')
plt.xlabel('Agency')
plt.ylabel('Number of Postings')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python

# ============================================================
# 7. THE "WON'T HIRE US" NARRATIVE
# ============================================================

# Look at entry-level or junior positions
# (You might need to adjust this filter based on actual column names)
entry_level_keywords = ['Junior', 'Entry', 'Assistant', 'Trainee', 'Intern']

# Try to find a relevant column — you can adjust this
if 'Civil Service Title' in df.columns:
    mask = df['Civil Service Title'].str.contains('|'.join(entry_level_keywords), case=False, na=False)
    entry_level = df[mask]
    print(f"🎯 Entry-level/junior postings: {len(entry_level)}")
else:
    print("⚠️ No 'Civil Service Title' column found — check your column names.")
```
